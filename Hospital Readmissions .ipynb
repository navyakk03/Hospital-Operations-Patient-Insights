```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
df
```

```python
df.isnull().sum()
```

```python
#Date formatting
df['admission_date'] = pd.to_datetime(df['admission_date'], dayfirst=True)
df['discharge_date'] = pd.to_datetime(df['discharge_date'], dayfirst=True)
```

```python
#validation
invalid_dates = df[df['discharge_date'] < df['admission_date']]
invalid_dates.shape
```

```python
df['calculated_los'] = (df['discharge_date'] - df['admission_date']).dt.days
(df['calculated_los'] == df['length_of_stay']).value_counts()
```

```python
df['length_of_stay'] = df['calculated_los']
df.drop(columns='calculated_los', inplace=True)
```

```python
df[['age', 'length_of_stay', 'total_cost', 'satisfaction']].describe()
```

```python
#Feature Engineering
```

```python
#Age Group
def age_group(age):
    if age < 18:
        return 'Child'
    elif age < 60:
        return 'Adult'
    else:
        return 'Elderly'

df['age_group'] = df['age'].apply(age_group)
```

```python
#Stay Category
df['stay_category'] = pd.cut(
    df['length_of_stay'],
    bins=[-1, 3, 7, 1000],
    labels=['Short Stay', 'Medium Stay', 'Long Stay']
)
```

```python
#Admission Year
df['admission_year'] = df['admission_date'].dt.year
```

```python
df[['age_group', 'stay_category', 'admission_year']].value_counts().head()#Validate engineered features
```
